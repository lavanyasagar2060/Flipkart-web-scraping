```python
#Importing an SKlearn and python library 
from sklearn.preprocessing import KBinsDiscretizer,OneHotEncoder,StandardScaler
import pandas as pd
```

```python
df["Price"]=df["Price"].str.replace("₹","")
df["Price"]=df["Price"].str.replace(",","")
df.head()
```

```python
#using bucketing to bin the price .

bins=[30000,35000,40000,45000,50000,55000]
label=["Low","Medium","High","Higher","Heighest"]
df["Price"]=pd.to_numeric(df["Price"])
df["price_bucket"]=pd.cut(df["Price"],bins=bins,labels=label)
df.head()
```

```python
#encodeing the data using one-hot encoder and scaled the data

oh=OneHotEncoder(sparse_output=False).set_output(transform ='pandas')
onehot=oh.fit_transform(df[["price_bucket"]])

ss=StandardScaler()
df["Price_standard_scaled"] = ss.fit_transform(df[["Price"]])
print(onehot.head())
print(df[["Price_standard_scaled"]].head())
```

```python
#scaling the data using Standard Scaler.

ss=StandardScaler()
df["Price_standard_scaled"] = ss.fit_transform(df[["Price"]])
df[["Price_standard_scaled"]].head()
```

```python
#Combining the features.

con_data = pd.concat([df, onehot, df[["Price_standard_scaled"]]],axis=1).drop(columns=["Title","Description","Reviews","price_bucket"], errors="ignore")
print(con_data.head())
```

```python
#using train_test_and split to train the data.
#Importing all the libraries.

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import accuracy_score
```

```python
#Dropping unnecessary columns:
x=con_data.drop(columns=["Price"],axis=1)
y=df[["Price"]]
```

```python
#Train-Test Split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

```python
# Shape of the model

print("Train_shape:",x_train.shape,y_train.shape)
print("Test_shape:",x_test.shape,y_test.shape)
```

```python
#Fitting data into the model.
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
#Predicting

y_pred=model.predict(x_test)
y_pred
```

```python
print("coefficients:",model.coef_)
print("Intercept:",model.intercept_)
```

```python
#Evaluating the regression model with R² and Mean Absolute Error .
from sklearn.metrics import r2_score,mean_absolute_error
print("Test_r2:",r2_score(y_test,y_pred))
print("Test_Mse:",mean_absolute_error(y_test,y_pred))
```

```python
#visualizing the regression performance.
plt.scatter(y_test,y_pred)
plt.plot([y_test.min(),y_test.max()],[y_test.min(),y_test.max()],"r--",label="perfect_fit")
plt.legend()
```
